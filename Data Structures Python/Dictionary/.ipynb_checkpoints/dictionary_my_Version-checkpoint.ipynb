```python
students = ['Arun', 'Ajay', 'Bob']
marks = [80, 95, 90]

student_marks = {}

for student, mark in zip(students, marks):
    student_marks[student] = mark 
```

```python
student_marks
```

```python
{student:marks for student, mark in zip(students, marks)}
```

```python
{student:marks for student, marks in zip(students, marks)}
```
