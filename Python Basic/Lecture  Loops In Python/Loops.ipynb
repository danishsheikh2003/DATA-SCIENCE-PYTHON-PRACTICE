```python
new_tuple = (1,4,9,16,25,36,49,64,81,100)
nums = 1
indx = 0
n = 49
while nums <= len(new_tuple):
    if new_tuple[indx] == n:
        print("Found at index",indx)
    print("Finding...",indx)    
    nums += 1
    indx += 1
print("Loops Ended")    
```
