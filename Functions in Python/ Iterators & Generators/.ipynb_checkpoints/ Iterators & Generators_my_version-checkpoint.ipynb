```python
#generator functions
#generator is a type of function which does not return a single value, instead return an iterator object 
```

```python
#regular function
def square_numbers(n):
    result= []
    for i in range(n):
        result.append(i ** 2) #astik astik
    return result 
```

```python
square_numbers(5)
```

```python
#yield 

def square_numbers(n):
    result= []
    for i in range(n):
        result.append(i ** 2) #astik astik
    yield i,result 
```

```python
square_numbers(5)
```

```python
gen = square_numbers(5)
```

```python
def square_numbers_generator(n):
    for i in range(n):
        yield i ** 2
```

```python
gen1 =square_numbers_generator(5)
```

```python
next(gen1)
```

```python
next(gen1)
```
