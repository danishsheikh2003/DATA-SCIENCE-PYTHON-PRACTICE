```python
def dashboard(func):
    
    def Wrapper():
        print("checking login...")
        func()
    return Wrapper

@dashboard
def welcome():
    print("Welcome to Dashboard")

welcome()
```

```python
def count(func):
    count = 0
    def Wrapper():
        nonlocal count

        count += 1 
        print(f"Function called {count} time")
        func()
    return Wrapper

@count
def calculate():
    print("Calculating...")

calculate()
calculate()
calculate()
```

```python
def addition(fun):

    def inside_the_addition(a,b):
        print("Adding Nuumber...")
        return fun(a,b)
        
    return inside_the_addition    
        
@addition
def add(a, b): 
    return a + b

print(add(12,13))  
```

```python
class MyDecorator:

    def __init__(self,func):
        self.func = func

    def __call__ (self):
        print("before")
        self.func
        print("After")
```

```python
@Mydecorator
def hello():
    print()
```
