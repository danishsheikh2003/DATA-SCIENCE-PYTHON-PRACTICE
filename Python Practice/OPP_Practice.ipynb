```python
# 🧪 Q2 — __init__() + Object Data
# Ab thoda next level.
# Ek Student class banao.
# Requirements:
# 1. __init__(self, name, age) banao.
# 2. name aur age ko object ke attributes mein store karo.
# 3. Ek method display() banao.
# 4. display() mein name aur age print karo.
# 5. Object banao:

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(self.name, self.age)

student = Student("Danish", 22)

print(student.age)
print(student.name)
student.display()
```

```python
# 🧪 Q3 — self ka asli test 🔥
# Ab dekhte hain self actually samjha ya sirf syntax yaad kiya hai. 😎
# Ek Employee class banao.
# Requirements:
# - __init__(self, name, salary) banao.
# - name aur salary ko object ke andar store karo.
# - show_details() method banao.
# - Method ke andar self.name aur self.salary use karke details print karo.
# - Do objects banao:


class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show_details(self):
        print(self.name, self.salary)

emp1 = Employee("Danish", 30000)
emp2 = Employee("Ajay", 40000)

emp1.show_details()
emp2.show_details()
```

```python
# 🧪 Q4 — Constructor + Calculation
# Ab ek Data Science-type practical question. 📊
# Dataset naam ki class banao.
# Requirements:
# 1. __init__(self, data) banao.
# 2. data ko self.data mein store karo.
# 3. get_count() naam ka method banao.
# 4. get_count() ko data ke total elements return karne chahiye.
# 5. Ye object banao:

class Dataset:

    def __init__ (self, data):
        self.data = data

    def get_count(self):
        return len(self.data)

    def get_average(self):
        return sum(self.data) / len(self.data)

dataset = Dataset([12, 13, 14, 15, 16, 17])

dataset.get_count()

dataset.get_average()
```

```python
# 🧪 Q6 — Encapsulation ka first test 🔐
# Ab Dataset class mein data ko private bana.
# Abhi:
# self.data = data


# isko private banana hai.
# Requirements:
# 1. self.__data use karo.
# 2. get_count() ko update karo.
# 3. get_average() ko update karo.
# 4. Ek naya method get_data() banao jo private data return kare.
# 5. Object banao:
# dataset = Dataset([10, 20, 30, 40, 50])


# Aur:
# print(dataset.get_data())
#print(dataset.get_count())
#print(dataset.get_average())


# Expected:
# [10, 20, 30, 40, 50]
# 5
# 30.0


class Dataset:

    def __init__ (self, data):
        self.__data = data

    def get_count(self):
        return len(self.__data)

    def get_average(self):
        return sum(self.__data) / len(self.__data)

    def get_data(self):
        return self.__data

dataset = Dataset([12, 13, 14, 15, 16, 17])

print(dataset.get_data())
print(dataset.get_count())
print(dataset.get_average())
```

```python
# 🧪 Q7 — Full OOP Mix 🔥
# Ab ek real Data Science-style mini problem.
# StudentDataset class banao.
# Requirements:
# 1. __init__(self, name, marks)
# - name ko store karo.
# - marks ko private rakho.
# 2. get_average()
# Marks ka average return kare.
# 3. get_max_marks()
# Highest marks return kare.
# 4. get_name()
# Student ka name return kare.
# Object:
# student = StudentDataset(    "Danish",    [78, 85, 92, 67, 88])


# Phir:
# print(student.get_name())print(student.get_average())print(student.get_max_marks())


# Expected:
# Danish
# 82.0
# 92

class StudentDataset:

    def __init__ (self, name, marks):
        self.name = name
        self.__marks = marks

    def get_average(self):
        return sum(self.__marks)/len(self.__marks)

    def get_max_marks(self):
        return max(self.__marks)

    def get_name(self):
        return self.name

student = StudentDataset("Danish",[89, 75, 99, 45, 49])    

print(student.get_name())
print(student.get_average())
print(student.get_max_marks())
```

```python
class FraudModel:
    
    def __init__ (self, model_name):
        self.__model_name = model_name

    def predict(self, transacion):
         print("General fraud prediction")

    def get_model_name(self):
        return self.__model_name

class LogisticRegression(FraudModel):
    
    def predict(self, transaction):
        print("Logistic Regression: Checking transaction")
        
        
class DecisionTree(FraudModel):
    
    def predict(self, transaction):
        print("Decision Tree: Checking transaction") 

logistic = LogisticRegression("Logistic Regression")
tree = DecisionTree("Decision Tree")

transaction = {
    "amount": 50000,
    "location": "Nagpur",
    "online": True
}


models = [logistic, tree]

for model in models:
    model.predict(transaction)
```

```python
# Question: Ek DataProcessor parent class bana.
# - __init__() → processor_name private rakho
# - process() → parent me "Processing data" print karo
# - PandasProcessor → process() override kare
# - NumpyProcessor → process() override kare
# - Dono ke objects banao
# - List me rakho
# - for loop se process() call karo

class DataProcessor:

    def __init__ (self, processor_name):
        self.__processor_name = processor_name

    def process(self):
        print("Processing data")

    def get_processor_name(self):
        print(self.__processor_name)


class PandaProcessor(DataProcessor):

    def process(self, process):
        print("Panda is proccesing data")


class NumpyProcessor(DataProcessor):

    def process(self, process):
        print("Numpy is proccesing data")


Panda = PandaProcessor("Panda")
Numpy = NumpyProcessor("Numpy")

process = { "data":50000 }

lab = [Panda, Numpy]

for i in lab:
    i.get_processor_name()
    i.process(process)
```

```python
# Requirements:
# 1. __init__(self, name) banao.
# 2. name ko private variable __name me store karo.
# 3. get_name() method se name return karo.
# 4. process(data) parent method banao.
# 5. Do child classes banao:
#    - DataCleaner
#    - FeatureEngineer
# 6. Dono child classes process() ko override karein.
# 7. Ye data use karo:
# data = [10, None, 20, None, 30]
# 8. Dono objects banao:
#    - "Data Cleaner"
#    - "Feature Engineer"
# 9. List me objects rakho.
# 10. Loop ke through:
#     - processor ka name print karo
#     - process(data) call karo.
# Expected output roughly:
# Data Cleaner
# Cleaning data

# Feature Engineer
# Creating features

class DataProcessor:
    
    def __init__(self, name):
        self.__name = name 

    def process(self, data):
        print("Data Processor")

    def get_name(self):
        print(self.__name)

class DataCleaner(DataProcessor):
    
    def process(self, data):
        print("Cleaning data")

class FeatureEngineer(DataProcessor):
    
    def process(self, data):
        print("Creating features")

Data    = DataCleaner("Data Cleaner")
Feature = FeatureEngineer("Feature Engineer")

data = [10, None, 20, None, 30]

Processing = [Data, Feature]

for i in Processing:
    i.get_name()
    i.process(data)   
    
```
