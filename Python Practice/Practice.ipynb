```python
# Q5 — Positive Even/Odd

# User se number input lo aur check karo:

# Number positive even hai
# Number positive odd hai
# Number negative even hai
# Number negative odd hai
# Number zero hai

Number = int(input("Enter The Number"))

if Number%2 == 0 and Number > 0 :
    print("This number is positive even")
if Number%2 == 0 and Number < 0:
    print("This number is negative even")
elif Number%2 != 0 and Number > 0:
    print("This number is positive Odd")
elif Number%2 != 0 and Number < 0: 
    print("This number is negative odd")
else:
    print("This Number is Equal to Zero")
    
```

```python
# Q6 — Simple Calculator

# User se:

# First number
# Second number
# Operator (+, -, *, /)

# input lo aur calculation perform karo.

User_num = int(input("Enter first Number"))
User_num1 = int(input("Enter Second Number"))

Operator = input(" +, -, *, /")

if Operator == "+":
    print(User_num + User_num1)
elif Operator == "-":
    print(User_num - User_num1) 
elif Operator == "*":
    print(User_num * User_num1) 
elif Operator == "/":
    print(User_num / User_num1)     
else:
    print("Operator error")
```
