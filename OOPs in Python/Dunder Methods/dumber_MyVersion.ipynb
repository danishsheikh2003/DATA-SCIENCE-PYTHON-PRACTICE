```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
```

```python
obj = Point(3,3)
obj1 = Point(3,3)
```

```python
print(obj == obj1)
```
