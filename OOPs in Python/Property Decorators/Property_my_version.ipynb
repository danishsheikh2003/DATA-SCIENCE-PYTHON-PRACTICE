```python
class Fruit:
    def __init__(self, name, price):
        self.__name = name
        self.__price = price

    @property                    #it will make access_price as attribute of the class
    def access_price(self):
        return self.__price
```

```python
class Fruit1:
    def __init__(self, name, price):
        self.__name = name
        self.__price = price

    @property                    
    def access_price(self):
        return self.__price

    @access_price.setter
    def price_set(self, price_new):
        self.__price = price_new
```

```python
fruit1 = Fruit1("Banana", 80)
```

```python
fruit1.access_price
```

```python
fruit1.price_set = 50000
```

```python
class Fruit3:
    def __init__(self, name, price):
        self.__name = name
        self.__price = price

    @property                    
    def access_price(self):
        return self.__price

    @access_price.setter
    def price_set(self, price_new):
        self.__price = price_new

    @access_price.deleter
    def del_price(self):
        del self.__price
```

```python
frt = Fruit3("Cherry", 700)
```

```python
frt.access_price
```

```python
frt.price_set = 200
```

```python
frt.access_price
```
