```python
class Calculetar:
    
    @staticmethod
    def add(x,y):
        return x+y
        
    @staticmethod
    def sub(x,y):
        return x-y  
        
    @staticmethod
    def milti(x,y):
        return x*y    
```

```python
Calculetar.add(12,13)
```

```python
class Bank3:
    def __init__(self, balance):
        self.balance = balance

    @classmethod
    def create_account(cls, balance):
        return cls(balance)

    def get_balance(self):
        return self.balance


bank3 = Bank3.create_account("Danish")

print(bank3.get_balance())
```
