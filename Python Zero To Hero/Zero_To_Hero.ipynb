```python
def calculate_average(numbers):
    
    total_sum = 0
    total_count = 0
    
    for i in numbers:
        total_sum = total_sum + i
        total_count += 1

    average = total_sum / total_count
    
    return average

calculate_average([10, 20, 30, 40])
```

```python
def count_passed(marks):

    count = 0 
    
    for i in marks:
        if i > 60:
            count += 1
            
    return count 

count_passed([45, 80, 67, 92, 55, 76])
```

```python
def find_max(number):

    largest = number[0]
    
    for i in number:
        if i > largest:
            largest = i

    return largest

print(find_max([10, 45, 23, 89, 34]))    
```

```python
students = {
    "Danish": 80,
    "Rahul": 55,
    "Sohel": 72,
    "Aman": 45
}

def get_average(students):
    
    total = 0
    count = 0

    for name, marks in students.items():
        if marks >= 60:
            total = total + marks
            count += 1

    average = total / count
    return average        
```

```python
get_average(students)
```

```python
def square_numbers(n):
    result= []
    for i in range(n):
        result.append(i ** 2) #astik astik
    yield i,result 
```
