```python
def calculate_average(numbers):
    
    total_sum = 0
    total_count = 0
    
    for i in numbers:
        total_sum = total_sum + i
        total_count += 1

    average = total_sum / total_count
    
    return average

calculate_average([10, 20, 30, 40])
```

```python
def count_passed(marks):

    count = 0 
    
    for i in marks:
        if i > 60:
            count += 1
            
    return count 

count_passed([45, 80, 67, 92, 55, 76])
```

```python
def find_max(number):

    largest = number[0]
    
    for i in number:
        if i > largest:
            largest = i

    return largest

print(find_max([10, 45, 23, 89, 34]))    
```
