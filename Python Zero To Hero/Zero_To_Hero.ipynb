```python
def calculate_average(numbers):
    
    total_sum = 0
    total_count = 0
    
    for i in numbers:
        total_sum = total_sum + i
        total_count += 1

    average = total_sum / total_count
    
    return average

calculate_average([10, 20, 30, 40])
```

```python
#Employee Salary System

class Employee:

    def __init__(self, name, salary):
        if salary < 0:
            raise ValueError("Salary cannot be negative")

        self.__name = name
        self.__salary = salary

    def display(self):
        print("Name: ",self.__name)
        print("Salary: ",self.__salary)

    def increase_salary(self, percent):
        if percent < 0:
            raise ValueError("Percentage cannot be negative")

        increase = self.__salary * percent / 100
        self.__salary += increase
        print("Salary increased successfully")

            
emp1 = Employee("Danish", 30000)

emp1.display()

emp1.increase_salary(10)

emp1.display()             
```

```python
class Employee:

    def __init__(self, name):
        self.name = name

    def display(self):
        print(f"Name: {self.name}")


class Developer(Employee):

    def __init__(self, name, language):
        super().__init__(name)
        self.language = language

    def show_language(self):
        print(f"Language: {self.language}")


employee = Developer("Danish", "Python")

employee.display()
employee.show_language()
```

```python
class Employee:

    def __init__(self, name, salary):
       self.name = name 
       self.salary = salary

class DataScientist(Employee):

    def __init__(self, name, salary, tool):
        super().__init__(name, salary)
        self.tool = tool 

    def display(self):
        print(f"Name: {self.name}\nSalary: {self.salary}")

    def show_tool(self):
        print(f"Tool: {self.tool}")

Show_Employee_Details = DataScientist("Danish",45000,"Python")

Show_Employee_Details.display()

Show_Employee_Details.show_tool()
```

```python
class Person:

    def __init__(self, name):
        self.name = name

    def display(self):
        print(f"Name: {self.name}")

class Student(Person):

    def __init__(self, name, course):
        super().__init__(name)
        self.course = course

    def display(self):
        super().display()
        print(f"Course: {self.course}")
        
student = Student("Danish","Python")

student.display()
        
```
