```python
#CSV>> comma separated values

data = [["Name", "course", "Fee"],
       ["Ajay", "DS", "1000"],
       ["Bijay", "ML", "2000"]]
```

```python
data
```

```python
for i in data:
    print(i)
```

```python
import csv 

with open("example_csv.csv", "w") as d:
   w = csv.writer(d)
   for i in data:
       w.writerow(i)
```

```python
with open("example_csv.csv", "r") as d:
   w = csv.reader(d)
   for i in data:
       print(i)
```

```python
#Json >> javscript object notation 
data = {"name": "Ajay",
       "Course": "DA",
       "fee":  2900}
```

```python
import json
with open("json2.json", "w") as f:
    json.dump(data, f)
```

```python
with open("json2.json", "r") as f:
    data = json.load(f)
print(data)    
```
